```python
import pandas as pd
import numpy as np
df = pd.read_csv('KaggleV2-May-2016.csv')
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().plot(kind='bar')
```

```python
df['Age'].fillna(df['Age'].median(),inplace=True)
```

```python
df['Neighbourhood'].fillna(df['Neighbourhood'].mode()[0], inplace=True)
```

```python
df.dropna(thresh=len(df)*0.8, inplace=True)
```

```python
df.isnull().sum()
```

```python
df.to_csv('cleaned_noshows.csv', index=False)
```
